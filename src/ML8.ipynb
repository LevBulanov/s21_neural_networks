```python
import pandas as pd

from MLP import MyMLP
from sklearn.neural_network import MLPClassifier

from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline 
from sklearn.model_selection import train_test_split

from sklearn.metrics import roc_auc_score

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import TensorDataset, DataLoader
```

## Загрузка и предобработка данных

```python
try:
    df = pd.read_csv('../datasets/training.csv')
except:
    df = pd.read_csv('/kaggle/input/ml5-data/training.csv')
```

```python
df = df.drop(['RefId'], axis=1)
df['PRIMEUNIT'] = df['PRIMEUNIT'].fillna('UNKNOWN')
df['AUCGUART'] = df['AUCGUART'].fillna('UNKNOWN')

df = df.drop('WheelTypeID', axis=1)
df['WheelType'] = df['WheelType'].fillna('UNKNOWN')
df['Trim'] = df['Trim'].fillna('UNKNOWN')

df = df.dropna()
```

### Разделение на выборки

```python
df['PurchDate'] = pd.to_datetime(df['PurchDate'])
df = df.sort_values(by='PurchDate', ascending=True)
train_val, test = train_test_split(df, shuffle=False, test_size=1/3)
train, val = train_test_split(train_val, shuffle=False, test_size=0.5)
del train_val
```

```python
train_end = '2009-09-15'
val_end = '2010-05-14'

train = df[df['PurchDate'] <= train_end]
val = df[(df['PurchDate'] > train_end) & (df['PurchDate'] <= val_end)]
test = df[df['PurchDate'] > val_end]
```

```python
train = train.drop('PurchDate', axis=1)
val = val.drop('PurchDate', axis=1)
test = test.drop('PurchDate', axis=1)
```

###  Кодирование фичей

```python
def build_preprocess_pipeline(train_df: pd.DataFrame):
    cat_cols = [c for c, dt in zip(train_df.columns, train_df.dtypes) if dt.kind not in ['i', 'f'] and c != 'IsBadBuy']
    num_cols = [c for c, dt in zip(train_df.columns, train_df.dtypes) if dt.kind in ['i', 'f'] and c != 'IsBadBuy']

    preproc = ColumnTransformer(
        transformers=[
            ('num', Pipeline(steps=[('scaler', StandardScaler())]), num_cols),
            ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
        ],
        remainder='drop'
    )
    return preproc
```

```python
def encode_features(preproc, df):
    X = pd.DataFrame(data=preproc.transform(df.drop('IsBadBuy', axis=1)),
                     columns=preproc.get_feature_names_out())
    y = df['IsBadBuy']
    return X, y
```

```python
preproc = build_preprocess_pipeline(train)
preproc.fit(train)
```

```python
X_train, y_train = encode_features(preproc, train)
X_val, y_val = encode_features(preproc, val)
X_test, y_test = encode_features(preproc, test)
```

## Обучение моделей

```python
def gini(actual, probs):
    auc = roc_auc_score(actual, probs)
    return 2 * auc - 1
```

### PyTorch

```python
class MLPTorch(nn.Module):
    def __init__(self, in_dim, hidden_dim, num_classes, activation='relu'):
        super().__init__()
        self.input_layer = nn.Linear(in_features=in_dim, out_features=hidden_dim)
        if activation == "relu":
            self.activation = nn.ReLU()
        elif activation == 'sigmoid':
            self.activation = nn.Sigmoid()
        elif activation == 'cosine':
            self.activation = lambda x: torch.cos(x)
        self.hidden_layer = nn.Linear(in_features=hidden_dim, out_features=num_classes)

    def forward(self, X):
        X = self.input_layer(X)
        X = self.activation(X)
        X = self.hidden_layer(X)
        return X.squeeze()
        
```

```python
features_train = torch.tensor(X_train.values, dtype=torch.float32)
features_val = torch.tensor(X_val.values, dtype=torch.float32)
features_test = torch.tensor(X_test.values, dtype=torch.float32)

target_train = torch.tensor(y_train.values, dtype=torch.float32)
target_val = torch.tensor(y_val.values, dtype=torch.float32)
target_test = torch.tensor(y_test.values, dtype=torch.float32)
```

```python
train_dataset = TensorDataset(features_train, target_train)
val_dataset = TensorDataset(features_val, target_val)
test_dataset = TensorDataset(features_test, target_test)
```

```python
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=True)
```

```python
num_epochs = 3

model = MLPTorch(in_dim=X_train.shape[1], hidden_dim=100, num_classes=1)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.BCEWithLogitsLoss()

for epoch in range(num_epochs):
    for batch_features, batch_labels in train_loader:
        optimizer.zero_grad()
        outputs = model(batch_features)
        loss = criterion(outputs, batch_labels)
        loss.backward()
        optimizer.step()
```

```python
model.eval()
with torch.no_grad():
    logits = model(features_train)
    probs = torch.sigmoid(logits).numpy()

gini(y_train, probs)
```

```python
model.eval()
with torch.no_grad():
    logits = model(features_val)
    probs = torch.sigmoid(logits).numpy()

gini(y_val, probs)
```
